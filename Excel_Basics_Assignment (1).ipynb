**Question 1. Explain the difference between Relative, Absolute and Mixed Cell Referencing.
You have the following dataset:
Vehicle Price
Car 500000
Bicycle 10500
Auto-Rickshaw 25000**

- **Difference Between Relative, Absolute, and Mixed Cell Referencing in Excel**

Given Dataset

| Vehicle       | Price (₹) |
| ------------- | --------- |
| Car           | 500000    |
| Bicycle       | 10500     |
| Auto-Rickshaw | 25000     |

 - 1. Relative Cell Referencing

Relative cell referencing is the default type of referencing in Excel. When a formula is copied to another cell, the reference automatically changes according to the new position.

Example:
Formula in cell C2:
=B2*10%
 - 2. Absolute Cell Referencing

Absolute cell referencing locks the cell reference so that it does not change when the formula is copied. This is done using the dollar sign ($).

Example:
=$B$2*10%

 - 3. Mixed Cell Referencing

Mixed cell referencing locks either the row or the column while allowing the other to change.

Examples:

$B2 → Column B is fixed, row changes

B$2 → Row 2 is fixed, column changes

**Comparison Table**

| Referencing Type | Example       | Behavior                   |
| ---------------- | ------------- | -------------------------- |
| Relative         | `B2`          | Row and column change      |
| Absolute         | `$B$2`        | Row and column fixed       |
| Mixed            | `$B2` / `B$2` | Either row or column fixed |

**Question 3. Using the data below, write a formula to calculate the average sales of items priced above 100 but less than
300:
Item Price Sales
Item A 90 1000
Item B 150 1200
Item C 250 1500**
 - **Formula to Calculate Average Sales of Items Priced Above 100 and Less Than 300**

 Given Data
 | Item   | Price (₹) | Sales |
| ------ | --------- | ----- |
| Item A | 90        | 1000  |
| Item B | 150       | 1200  |
| Item C | 250       | 1500  |

=AVERAGEIFS(C2:C4, B2:B4, ">100", B2:B4, "<300")

Explanation

AVERAGEIFS calculates the average of values in the Sales column.

C2:C4 → Range containing sales values.

B2:B4 → Range containing prices.

">100" and "<300" → Conditions applied to prices.
